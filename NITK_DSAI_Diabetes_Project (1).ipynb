```python

# =============================================================================
# NITK SURATHKAL | B.TECH YEAR 1 | DSAI FINAL COURSE PROJECT
# DIABETES RISK CLASSIFICATION AND EXPLORATORY DATA ANALYSIS
#
# Dataset: Diabetes Prediction Dataset (100,000 rows)
# Source: Kaggle - iammustafatz/diabetes-prediction-dataset
#
# Course concepts demonstrated:
# 1. Data import, inspection, data types and descriptive statistics
# 2. Data cleaning, missing values, duplicates and outlier analysis
# 3. Encoding, normalization, standardization and feature selection
# 4. Histograms, bar charts, boxplots, scatterplots and correlation
# 5. Supervised classification: Logistic Regression, KNN, Decision Tree,
#    Random Forest and Gradient Boosting
# 6. Neural-network classification using an MLP (optional comparison)
# 7. Model evaluation, confusion matrix, precision, recall, F1, F2,
#    ROC-AUC and PR-AUC
# 8. Unsupervised learning: K-Means clustering and WCSS
# 9. Dimensionality reduction using PCA
# 10. Model interpretation, fairness checks and limitations
#
# IMPORTANT:
# This is an educational screening project, NOT a medical diagnostic system.
# =============================================================================


# =============================================================================
# 0. CONFIGURATION
# =============================================================================

FORCE_RETRAIN = False       # Set True to train models again from scratch
RANDOM_STATE = 42
TEST_SIZE = 0.15
VALIDATION_SIZE = 0.15
CV_FOLDS = 3
KNN_MAX_ROWS = 25000
MLP_MAX_ROWS = 40000

# Set to True if you want to include the neural-network model.
# It can increase execution time.
INCLUDE_NEURAL_NETWORK = True

# Kaggle dataset identifier
KAGGLE_DATASET = "iammustafatz/diabetes-prediction-dataset"
CANONICAL_CSV_NAME = "diabetes_prediction_dataset.csv"


# =============================================================================
# 1. INSTALL AND IMPORT LIBRARIES
# =============================================================================

import sys
import subprocess
import importlib.util

def install_if_missing(package, import_name=None):
    import_name = import_name or package
    if importlib.util.find_spec(import_name) is None:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", "-q", package]
        )

install_if_missing("kagglehub")
install_if_missing("ipywidgets")

import os
import glob
import json
import time
import hashlib
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown, clear_output

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RandomizedSearchCV
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    OneHotEncoder
)
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    HistGradientBoostingClassifier
)
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)
from sklearn.inspection import permutation_importance
from sklearn.decomposition import PCA
from sklearn.cluster import MiniBatchKMeans
from sklearn.exceptions import ConvergenceWarning

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=ConvergenceWarning)

np.random.seed(RANDOM_STATE)

print("Libraries imported successfully.")


# =============================================================================
# 2. CONNECT GOOGLE DRIVE AND CREATE PROJECT FOLDERS
# =============================================================================

from google.colab import drive
drive.mount("/content/drive")

PROJECT_DIR = "/content/drive/MyDrive/NITK_DSAI_Diabetes_Project"

DATA_DIR = os.path.join(PROJECT_DIR, "Dataset")
MODEL_DIR = os.path.join(PROJECT_DIR, "Models")
OUTPUT_DIR = os.path.join(PROJECT_DIR, "Outputs")
PLOT_DIR = os.path.join(OUTPUT_DIR, "Visualizations")

for folder in [PROJECT_DIR, DATA_DIR, MODEL_DIR, OUTPUT_DIR, PLOT_DIR]:
    os.makedirs(folder, exist_ok=True)

CSV_PATH = os.path.join(DATA_DIR, CANONICAL_CSV_NAME)

MODEL_PATH = os.path.join(MODEL_DIR, "best_diabetes_model.joblib")
METADATA_PATH = os.path.join(MODEL_DIR, "model_metadata.json")

CLEANED_DATA_PATH = os.path.join(DATA_DIR, "cleaned_diabetes_dataset.csv")
METRICS_PATH = os.path.join(OUTPUT_DIR, "model_metrics.csv")
COMPARISON_PATH = os.path.join(OUTPUT_DIR, "model_comparison.csv")
FEATURE_IMPORTANCE_PATH = os.path.join(
    OUTPUT_DIR, "feature_importance.csv"
)
SUMMARY_PATH = os.path.join(OUTPUT_DIR, "project_summary.json")

print("\nProject folders ready:")
print(PROJECT_DIR)


# =============================================================================
# 3. DOWNLOAD DATASET ONLY IF THE ORIGINAL IS NOT ALREADY IN DRIVE
# =============================================================================

if not os.path.exists(CSV_PATH):

    print("\nOriginal dataset not found in Drive.")
    print("Downloading the Kaggle dataset...")

    import kagglehub
    downloaded_folder = kagglehub.dataset_download(KAGGLE_DATASET)

    downloaded_csvs = glob.glob(
        os.path.join(downloaded_folder, "**", "*.csv"),
        recursive=True
    )

    if not downloaded_csvs:
        raise FileNotFoundError(
            "No CSV file was found in the downloaded Kaggle dataset."
        )

    # Prefer the expected original dataset filename.
    matching = [
        p for p in downloaded_csvs
        if os.path.basename(p).lower() == CANONICAL_CSV_NAME.lower()
    ]

    source_csv = matching[0] if matching else downloaded_csvs[0]

    import shutil
    shutil.copy2(source_csv, CSV_PATH)

    print("Original dataset saved permanently in Google Drive.")

else:
    print("\nOriginal dataset already exists in Drive.")
    print("Using saved original dataset. No download required.")

print("Dataset path:", CSV_PATH)


# =============================================================================
# 4. LOAD AND INSPECT THE ORIGINAL DATASET
# =============================================================================

df_original = pd.read_csv(CSV_PATH)

print("\n" + "=" * 75)
print("INITIAL DATASET INSPECTION")
print("=" * 75)

print("\nDataset shape:", df_original.shape)

print("\nFirst five rows:")
display(df_original.head())

print("\nLast five rows:")
display(df_original.tail())

print("\nColumn names:")
print(df_original.columns.tolist())

print("\nData types:")
display(df_original.dtypes.to_frame("Data Type"))

print("\nDataset information:")
df_original.info()

print("\nDescriptive statistics:")
display(df_original.describe(include="all").T)

print("\nMissing values:")
display(
    df_original.isnull().sum()
    .to_frame("Missing Values")
    .assign(
        Missing_Percentage=lambda x:
        100 * x["Missing Values"] / len(df_original)
    )
)

print("\nNumber of exact duplicate rows:")
print(df_original.duplicated().sum())

print("\nUnique values in categorical columns:")
for col in df_original.select_dtypes(include=["object", "category"]).columns:
    print(f"\n{col}:")
    print(df_original[col].value_counts(dropna=False))

print("\nUnique values in binary / target columns:")
for col in ["hypertension", "heart_disease", "diabetes"]:
    if col in df_original.columns:
        print(f"\n{col}:")
        print(df_original[col].value_counts(dropna=False))


# =============================================================================
# 5. VERIFY DATASET STRUCTURE
# =============================================================================

REQUIRED_COLUMNS = [
    "gender",
    "age",
    "hypertension",
    "heart_disease",
    "smoking_history",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level",
    "diabetes"
]

missing_columns = [
    col for col in REQUIRED_COLUMNS
    if col not in df_original.columns
]

if missing_columns:
    raise ValueError(
        f"The dataset is missing required columns: {missing_columns}"
    )

df = df_original[REQUIRED_COLUMNS].copy()

print("\nRequired dataset columns verified.")


# =============================================================================
# 6. INITIAL TARGET DISTRIBUTION
# =============================================================================

print("\nOriginal target distribution:")
display(df["diabetes"].value_counts(dropna=False).to_frame("Count"))

plt.figure(figsize=(7, 5))
df["diabetes"].value_counts().sort_index().plot(
    kind="bar",
    edgecolor="black"
)
plt.title("Original Diabetes Class Distribution")
plt.xlabel("Diabetes (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(
    os.path.join(PLOT_DIR, "01_original_class_distribution.png"),
    dpi=150
)
plt.show()


# =============================================================================
# 7. DATA QUALITY INVESTIGATION BEFORE CLEANING
# =============================================================================

print("\n" + "=" * 75)
print("DATA QUALITY INVESTIGATION")
print("=" * 75)

print("\nGender categories:")
print(df["gender"].value_counts(dropna=False))

print("\nSmoking history categories:")
print(df["smoking_history"].value_counts(dropna=False))

print("\nZero values in measurements:")
for col in ["age", "bmi", "HbA1c_level", "blood_glucose_level"]:
    print(f"{col}: {(df[col] == 0).sum()}")

print("\nOutlier investigation using the IQR method:")

numeric_investigation = [
    "age",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level"
]

outlier_rows = []

for col in numeric_investigation:

    values = pd.to_numeric(df[col], errors="coerce").dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = values[
        (values < lower_bound) | (values > upper_bound)
    ]

    outlier_rows.append({
        "Feature": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": 100 * len(outliers) / len(values)
    })

outlier_report = pd.DataFrame(outlier_rows)
display(outlier_report)

# Important:
# IQR is used to identify potential outliers, not automatically delete them.
# Extreme BMI, glucose or HbA1c values may be genuine observations.
# Removing them without evidence could remove important high-risk cases.


# =============================================================================
# 8. DATA CLEANING
# =============================================================================

print("\n" + "=" * 75)
print("DATA CLEANING")
print("=" * 75)

rows_before = len(df)

# 8.1 Remove rows with an invalid / unspecified gender category.
df["gender"] = df["gender"].astype("string").str.strip()
df["gender"] = df["gender"].str.title()

other_gender_count = (df["gender"] == "Other").sum()
df = df[df["gender"] != "Other"].copy()

print("Rows removed due to gender = Other:", other_gender_count)

# 8.2 Standardize smoking categories.
df["smoking_history"] = (
    df["smoking_history"]
    .astype("string")
    .str.strip()
    .str.lower()
)

smoking_mapping = {
    "ever": "former",
    "not current": "former",
    "former": "former",
    "never": "never",
    "current": "current",
    "no info": np.nan,
    "no_info": np.nan
}

df["smoking_history"] = (
    df["smoking_history"]
    .map(smoking_mapping)
    .astype("object")
)

# 8.3 Convert numeric columns safely.
numeric_columns = [
    "age",
    "hypertension",
    "heart_disease",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level",
    "diabetes"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 8.4 Project data-quality rule:
# Zero values in these measurement fields are treated as missing.
# Note: age = 0 can be a genuine infant age. This is a project-specific
# assumption and must be acknowledged in the report.
zero_as_missing = [
    "age",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level"
]

for col in zero_as_missing:
    df.loc[df[col] == 0, col] = np.nan

# 8.5 Target must be known.
df = df.dropna(subset=["diabetes"]).copy()

# Retain only valid binary target values.
df = df[df["diabetes"].isin([0, 1])].copy()

# 8.6 Validate binary predictor columns.
for col in ["hypertension", "heart_disease"]:
    df.loc[~df[col].isin([0, 1]), col] = np.nan

# 8.7 Remove exact duplicates after standardization.
duplicates_before_removal = df.duplicated().sum()
df = df.drop_duplicates().copy()

# 8.8 Convert binary columns to conventional integer types.
for col in ["hypertension", "heart_disease", "diabetes"]:
    df[col] = df[col].astype(int)

# 8.9 Normalize categorical text.
df["gender"] = df["gender"].astype("object")
df["smoking_history"] = df["smoking_history"].astype("object")

df.reset_index(drop=True, inplace=True)

print("\nCleaning summary:")
print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", len(df))
print("Exact duplicate rows removed:", duplicates_before_removal)
print("Rows removed overall:", rows_before - len(df))

print("\nMissing values after cleaning:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nCleaned data preview:")
display(df.head())

print("\nCleaned data types:")
display(df.dtypes.to_frame("Data Type"))

print("\nCleaned target distribution:")
display(df["diabetes"].value_counts().to_frame("Count"))

df.to_csv(CLEANED_DATA_PATH, index=False)
print("\nCleaned dataset saved to Drive:", CLEANED_DATA_PATH)


# =============================================================================
# 9. INVESTIGATE IDENTICAL FEATURE PROFILES WITH CONFLICTING LABELS
# =============================================================================

# Repeated patient-like profiles can otherwise appear in both train and test.
# We group identical predictor profiles and keep them in one split.
# If identical predictors have conflicting target labels, the profile is
# ambiguous and is excluded rather than arbitrarily assigning a label.

FEATURE_COLUMNS = [
    "gender",
    "age",
    "hypertension",
    "heart_disease",
    "smoking_history",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level"
]

df["_profile_id"] = pd.util.hash_pandas_object(
    df[FEATURE_COLUMNS],
    index=False
).astype("uint64")

profile_label_counts = (
    df.groupby("_profile_id")["diabetes"].nunique()
)

conflicting_profiles = profile_label_counts[
    profile_label_counts > 1
].index

conflicting_rows = df["_profile_id"].isin(conflicting_profiles).sum()

print("\nIdentical predictor profiles with conflicting labels:",
      len(conflicting_profiles))
print("Rows belonging to conflicting profiles:", conflicting_rows)

if len(conflicting_profiles) > 0:
    df = df[
        ~df["_profile_id"].isin(conflicting_profiles)
    ].copy()

print("Rows remaining after ambiguous-profile investigation:", len(df))


# =============================================================================
# 10. TRAIN / VALIDATION / TEST SPLIT WITHOUT PROFILE LEAKAGE
# =============================================================================

# First split unique feature profiles, then assign all rows of each profile
# to the same partition.

profile_table = (
    df.groupby("_profile_id", as_index=False)["diabetes"]
    .first()
)

all_groups = profile_table["_profile_id"].values
group_labels = profile_table["diabetes"].values

train_groups, remaining_groups = train_test_split(
    all_groups,
    test_size=TEST_SIZE + VALIDATION_SIZE,
    random_state=RANDOM_STATE,
    stratify=group_labels
)

remaining_label_map = (
    profile_table.set_index("_profile_id")["diabetes"]
)

remaining_labels = np.array([
    remaining_label_map.loc[g] for g in remaining_groups
])

validation_fraction_of_remaining = (
    VALIDATION_SIZE / (TEST_SIZE + VALIDATION_SIZE)
)

validation_groups, test_groups = train_test_split(
    remaining_groups,
    test_size=1 - validation_fraction_of_remaining,
    random_state=RANDOM_STATE,
    stratify=remaining_labels
)

train_df = df[df["_profile_id"].isin(train_groups)].copy()
val_df = df[df["_profile_id"].isin(validation_groups)].copy()
test_df = df[df["_profile_id"].isin(test_groups)].copy()

# Confirm that no identical feature profile crosses partitions.
assert set(train_groups).isdisjoint(set(validation_groups))
assert set(train_groups).isdisjoint(set(test_groups))
assert set(validation_groups).isdisjoint(set(test_groups))

X_train = train_df[FEATURE_COLUMNS].copy()
y_train = train_df["diabetes"].copy()

X_val = val_df[FEATURE_COLUMNS].copy()
y_val = val_df["diabetes"].copy()

X_test = test_df[FEATURE_COLUMNS].copy()
y_test = test_df["diabetes"].copy()

print("\nData split:")
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print("\nPositive class percentage:")
print("Train:", round(100 * y_train.mean(), 2), "%")
print("Validation:", round(100 * y_val.mean(), 2), "%")
print("Test:", round(100 * y_test.mean(), 2), "%")

print("\nThe test set remains separate from model selection and threshold tuning.")


# =============================================================================
# 11. EXPLORATORY DATA ANALYSIS
# =============================================================================

print("\n" + "=" * 75)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 75)

eda_df = df.drop(columns=["_profile_id"]).copy()

# 11.1 Missing-value visualization.
plt.figure(figsize=(10, 5))
missing_percent = (
    eda_df.isnull().mean().sort_values(ascending=False) * 100
)
missing_percent.plot(kind="bar", edgecolor="black")
plt.title("Missing Values by Feature")
plt.xlabel("Feature")
plt.ylabel("Missing Values (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "02_missing_values.png"), dpi=150)
plt.show()

# 11.2 Numerical distributions.
for col in numeric_investigation:

    plt.figure(figsize=(8, 5))
    eda_df[col].dropna().hist(bins=35, edgecolor="black")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.savefig(
        os.path.join(PLOT_DIR, f"03_distribution_{col}.png"),
        dpi=150
    )
    plt.show()

# 11.3 Boxplots.
for col in numeric_investigation:

    plt.figure(figsize=(8, 4))
    plt.boxplot(eda_df[col].dropna(), vert=False)
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.savefig(
        os.path.join(PLOT_DIR, f"04_boxplot_{col}.png"),
        dpi=150
    )
    plt.show()

# 11.4 Categorical distributions.
for col in ["gender", "smoking_history"]:

    plt.figure(figsize=(8, 5))
    eda_df[col].value_counts(dropna=False).plot(
        kind="bar",
        edgecolor="black"
    )
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.savefig(
        os.path.join(PLOT_DIR, f"05_category_{col}.png"),
        dpi=150
    )
    plt.show()

# 11.5 Diabetes prevalence by category.
for col in ["gender", "smoking_history"]:

    rate = (
        eda_df.groupby(col, dropna=False)["diabetes"]
        .mean()
        .sort_values(ascending=False) * 100
    )

    plt.figure(figsize=(8, 5))
    rate.plot(kind="bar", edgecolor="black")
    plt.title(f"Diabetes Rate by {col}")
    plt.xlabel(col)
    plt.ylabel("Diabetes Rate (%)")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.savefig(
        os.path.join(PLOT_DIR, f"06_diabetes_rate_{col}.png"),
        dpi=150
    )
    plt.show()

# 11.6 Numerical feature distributions by diabetes class.
for col in numeric_investigation:

    plt.figure(figsize=(8, 5))

    for target_value, label in [(0, "No Diabetes"), (1, "Diabetes")]:
        values = eda_df.loc[
            eda_df["diabetes"] == target_value, col
        ].dropna()

        plt.hist(
            values,
            bins=30,
            alpha=0.55,
            label=label
        )

    plt.title(f"{col} Distribution by Diabetes Class")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(PLOT_DIR, f"07_{col}_by_diabetes.png"),
        dpi=150
    )
    plt.show()

# 11.7 Scatterplot.
scatter_sample = eda_df.sample(
    min(5000, len(eda_df)),
    random_state=RANDOM_STATE
)

plt.figure(figsize=(8, 6))
for target_value, label in [(0, "No Diabetes"), (1, "Diabetes")]:
    subset = scatter_sample[
        scatter_sample["diabetes"] == target_value
    ]
    plt.scatter(
        subset["HbA1c_level"],
        subset["blood_glucose_level"],
        alpha=0.35,
        s=12,
        label=label
    )

plt.title("HbA1c Level vs Blood Glucose Level")
plt.xlabel("HbA1c Level")
plt.ylabel("Blood Glucose Level")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "08_hba1c_vs_glucose.png"), dpi=150)
plt.show()

# 11.8 Correlation matrix

# Select all numerical columns, including the binary diabetes target.
correlation_columns = [
    "age",
    "hypertension",
    "heart_disease",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level",
    "diabetes"
]
# Calculate Pearson correlation coefficients.
correlation = eda_df[correlation_columns].corr(method="pearson")
# Create the heatmap.
fig, ax = plt.subplots(figsize=(11, 9))
heatmap = ax.imshow(
    correlation,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    aspect="equal"
)

# Add a colour bar with a clear label.
cbar = fig.colorbar(heatmap, ax=ax, shrink=0.85)
cbar.set_label("Pearson Correlation Coefficient", fontsize=11)
# Label the horizontal and vertical axes.
ax.set_xticks(np.arange(len(correlation_columns)))
ax.set_yticks(np.arange(len(correlation_columns)))
ax.set_xticklabels(
    correlation_columns,
    rotation=40,
    ha="right",
    fontsize=10
)
ax.set_yticklabels(
    correlation_columns,
    fontsize=10
)
# Write the exact correlation coefficient inside every square.
for i in range(len(correlation_columns)):
    for j in range(len(correlation_columns)):
        value = correlation.iloc[i, j]
        # Use white text on stronger-coloured squares for readability.
        text_color = "white" if abs(value) >= 0.55 else "black"
        ax.text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center",
            color=text_color,
            fontsize=10,
            fontweight="bold"
        )

# Add white borders between individual squares.
ax.set_xticks(
    np.arange(-0.5, len(correlation_columns), 1),
    minor=True
)
ax.set_yticks(
    np.arange(-0.5, len(correlation_columns), 1),
    minor=True
)
ax.grid(
    which="minor",
    color="white",
    linestyle="-",
    linewidth=1.5
)
ax.tick_params(which="minor", bottom=False, left=False)
ax.set_title(
    "Correlation Heatmap of Numerical Features and Diabetes",
    fontsize=14,
    fontweight="bold",
    pad=18
)
plt.tight_layout()
plt.savefig(
    os.path.join(PLOT_DIR, "09_correlation_matrix.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# Display each feature's correlation with the diabetes target,
# ordered from strongest positive to weakest.
print("\nCorrelation of each numerical feature with diabetes:")

display(
    correlation["diabetes"]
    .drop("diabetes")
    .sort_values(ascending=False)
    .to_frame("Pearson Correlation with Diabetes")
)

# =============================================================================
# 12. PREPROCESSING: IMPUTATION, ENCODING AND SCALING
# =============================================================================

NUMERIC_FEATURES = [
    "age",
    "hypertension",
    "heart_disease",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level"
]

CATEGORICAL_FEATURES = [
    "gender",
    "smoking_history"
]

# All imputers and scalers are fitted inside the training pipeline
# This prevents information from validation/test sets leaking into training

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(
        strategy="median",
        add_indicator=True
    )),
    ("scaler", StandardScaler())
])

# Compatibility with older and newer scikit-learn versions
try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(
        strategy="most_frequent",
        add_indicator=True
    )),
    ("encoder", encoder)
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, NUMERIC_FEATURES),
        ("categorical", categorical_pipeline, CATEGORICAL_FEATURES)
    ],
    remainder="drop"
)

print("\nPreprocessing pipeline created.")
print("Numerical features:", NUMERIC_FEATURES)
print("Categorical features:", CATEGORICAL_FEATURES)


# =============================================================================
# 13. DEMONSTRATE STANDARDIZATION AND MIN-MAX NORMALIZATION
# =============================================================================

# Demonstration on training data only
# These separate demonstrations are not used to fit the final model

standard_scaler_demo = StandardScaler()
minmax_scaler_demo = MinMaxScaler()

standardized_train = standard_scaler_demo.fit_transform(
    X_train[NUMERIC_FEATURES]
)

minmax_train = minmax_scaler_demo.fit_transform(
    X_train[NUMERIC_FEATURES]
)

print("\nScaling demonstration:")
print("Original numerical data:")
display(X_train[NUMERIC_FEATURES].head())

print("Standardized data (first five rows):")
display(pd.DataFrame(
    standardized_train[:5],
    columns=NUMERIC_FEATURES
))

print("Min-Max normalized data (first five rows):")
display(pd.DataFrame(
    minmax_train[:5],
    columns=NUMERIC_FEATURES
))


# =============================================================================
# 14. FEATURE SELECTION / FEATURE IMPORTANCE
# =============================================================================

# The final model uses the available project features
# Permutation importance later evaluates how much each original feature
# contributes to predictive performance
# We're not selecting features using the test set


# =============================================================================
# 15. DATA AND CONFIGURATION FINGERPRINT FOR MODEL CACHING
# =============================================================================

data_hash = hashlib.sha256(
    pd.util.hash_pandas_object(
        df.drop(columns=["_profile_id"]),
        index=False
    ).values.tobytes()
).hexdigest()

config = {
    "random_state": RANDOM_STATE,
    "cv_folds": CV_FOLDS,
    "features": FEATURE_COLUMNS,
    "include_neural_network": INCLUDE_NEURAL_NETWORK,
    "test_size": TEST_SIZE,
    "validation_size": VALIDATION_SIZE
}

config_hash = hashlib.sha256(
    json.dumps(config, sort_keys=True).encode()
).hexdigest()

fingerprint = hashlib.sha256(
    (data_hash + config_hash).encode()
).hexdigest()

print("\nDataset/configuration fingerprint:", fingerprint[:16])


# =============================================================================
# 16. MODEL TRAINING AND HYPERPARAMETER TUNING
# =============================================================================

best_model = None
best_model_name = None
best_threshold = 0.5
model_comparison = []
cached_model_loaded = False

if (
    os.path.exists(MODEL_PATH)
    and os.path.exists(METADATA_PATH)
    and not FORCE_RETRAIN
):

    try:
        with open(METADATA_PATH, "r") as f:
            saved_metadata = json.load(f)

        if saved_metadata.get("fingerprint") == fingerprint:

            best_model = joblib.load(MODEL_PATH)
            best_model_name = saved_metadata["best_model_name"]
            best_threshold = float(
                saved_metadata.get("best_threshold", 0.5)
            )

            model_comparison = saved_metadata.get(
                "model_comparison", []
            )

            cached_model_loaded = True

            print("\n" + "=" * 75)
            print("SAVED MODEL SUCCESSFULLY LOADED")
            print("=" * 75)
            print("Model:", best_model_name)
            print("Saved threshold:", best_threshold)
            print("No model retraining required.")

        else:
            print("\nSaved model belongs to a different dataset/configuration.")
            print("Retraining is required.")

    except Exception as e:
        print("\nCould not load cached model:", repr(e))
        print("A fresh training run will be performed.")

if FORCE_RETRAIN:
    print("\nFORCE_RETRAIN=True. Training from scratch.")


if not cached_model_loaded:

    print("\n" + "=" * 75)
    print("MODEL TRAINING")
    print("=" * 75)

    cv = StratifiedKFold(
        n_splits=CV_FOLDS,
        shuffle=True,
        random_state=RANDOM_STATE
    )

    # Main models.
    model_definitions = {
        "Logistic Regression": (
            LogisticRegression(
                max_iter=1200,
                class_weight="balanced",
                random_state=RANDOM_STATE
            ),
            {
                "model__C": [0.05, 0.1, 0.5, 1, 2, 5, 10],
                "model__solver": ["lbfgs"]
            }
        ),

        "K-Nearest Neighbours": (
            KNeighborsClassifier(),
            {
                "model__n_neighbors": [5, 9, 15, 21],
                "model__weights": ["uniform", "distance"],
                "model__p": [1, 2]
            }
        ),

        "Decision Tree": (
            DecisionTreeClassifier(
                class_weight="balanced",
                random_state=RANDOM_STATE
            ),
            {
                "model__max_depth": [4, 6, 8, 12, None],
                "model__min_samples_split": [2, 10, 25],
                "model__min_samples_leaf": [1, 5, 10]
            }
        ),

        "Random Forest": (
            RandomForestClassifier(
                n_estimators=180,
                class_weight="balanced_subsample",
                random_state=RANDOM_STATE,
                n_jobs=-1
            ),
            {
                "model__max_depth": [8, 12, 18, None],
                "model__min_samples_split": [2, 10],
                "model__min_samples_leaf": [1, 3, 5],
                "model__max_features": ["sqrt", 0.8]
            }
        ),

        "Gradient Boosting": (
            GradientBoostingClassifier(
                random_state=RANDOM_STATE
            ),
            {
                "model__n_estimators": [80, 120, 160],
                "model__learning_rate": [0.03, 0.05, 0.1],
                "model__max_depth": [2, 3]
            }
        )
    }

    # KNN is computationally expensive on the full dataset.
    # Use a reproducible stratified subset for its model fitting.
    X_knn = X_train
    y_knn = y_train

    if len(X_train) > KNN_MAX_ROWS:

        X_knn, _, y_knn, _ = train_test_split(
            X_train,
            y_train,
            train_size=KNN_MAX_ROWS,
            random_state=RANDOM_STATE,
            stratify=y_train
        )

        print(
            f"\nKNN training subset: {len(X_knn):,} rows "
            f"(full training set: {len(X_train):,})"
        )

    # Train each tuned model.
    for model_name, (estimator, parameter_grid) in model_definitions.items():

        print("\n" + "-" * 65)
        print("Training:", model_name)
        print("-" * 65)

        model_pipeline = Pipeline(steps=[
            ("prep", joblib.load if False else preprocessor),
            ("model", estimator)
        ])

        if model_name == "K-Nearest Neighbours":
            fit_X, fit_y = X_knn, y_knn
        else:
            fit_X, fit_y = X_train, y_train

        search = RandomizedSearchCV(
            estimator=model_pipeline,
            param_distributions=parameter_grid,
            n_iter=3,
            scoring="recall",
            cv=cv,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            verbose=0,
            refit=True,
            error_score="raise"
        )

        start_time = time.time()
        search.fit(fit_X, fit_y)
        elapsed = time.time() - start_time

        fitted_model = search.best_estimator_

        val_probabilities = fitted_model.predict_proba(X_val)[:, 1]

        val_predictions_05 = (
            val_probabilities >= 0.5
        ).astype(int)

        model_comparison.append({
            "Model": model_name,
            "CV Best Recall": search.best_score_,
            "Validation Recall at 0.5": recall_score(
                y_val, val_predictions_05, zero_division=0
            ),
            "Validation Precision at 0.5": precision_score(
                y_val, val_predictions_05, zero_division=0
            ),
            "Validation F1 at 0.5": f1_score(
                y_val, val_predictions_05, zero_division=0
            ),
            "Validation F2 at 0.5": fbeta_score(
                y_val, val_predictions_05, beta=2, zero_division=0
            ),
            "Validation PR-AUC": average_precision_score(
                y_val, val_probabilities
            ),
            "Training Time (seconds)": round(elapsed, 2),
            "Best Parameters": str(search.best_params_)
        })

        # Select by validation F2 at the default threshold initially.
        current_f2 = fbeta_score(
            y_val,
            val_predictions_05,
            beta=2,
            zero_division=0
        )

        if best_model is None or current_f2 > best_model_selection_score:
            best_model = fitted_model
            best_model_name = model_name
            best_model_selection_score = current_f2

        print("Best CV recall:", round(search.best_score_, 4))
        print("Validation F2 at 0.5:", round(current_f2, 4))
        print("Training time:", round(elapsed, 2), "seconds")

    # Optional neural-network comparison.
    if INCLUDE_NEURAL_NETWORK:

        print("\n" + "-" * 65)
        print("Training: Neural Network (MLPClassifier)")
        print("-" * 65)

        X_mlp = X_train
        y_mlp = y_train

        if len(X_train) > MLP_MAX_ROWS:
            X_mlp, _, y_mlp, _ = train_test_split(
                X_train,
                y_train,
                train_size=MLP_MAX_ROWS,
                random_state=RANDOM_STATE,
                stratify=y_train
            )

        mlp_pipeline = Pipeline(steps=[
            ("prep", preprocessor),
            ("model", MLPClassifier(
                hidden_layer_sizes=(32, 16),
                activation="relu",
                solver="adam",
                alpha=0.001,
                learning_rate_init=0.001,
                max_iter=100,
                early_stopping=True,
                random_state=RANDOM_STATE
            ))
        ])

        start_time = time.time()
        mlp_pipeline.fit(X_mlp, y_mlp)
        elapsed = time.time() - start_time

        mlp_probabilities = mlp_pipeline.predict_proba(X_val)[:, 1]
        mlp_predictions = (mlp_probabilities >= 0.5).astype(int)

        mlp_f2 = fbeta_score(
            y_val, mlp_predictions, beta=2, zero_division=0
        )

        model_comparison.append({
            "Model": "Neural Network (MLP)",
            "CV Best Recall": np.nan,
            "Validation Recall at 0.5": recall_score(
                y_val, mlp_predictions, zero_division=0
            ),
            "Validation Precision at 0.5": precision_score(
                y_val, mlp_predictions, zero_division=0
            ),
            "Validation F1 at 0.5": f1_score(
                y_val, mlp_predictions, zero_division=0
            ),
            "Validation F2 at 0.5": mlp_f2,
            "Validation PR-AUC": average_precision_score(
                y_val, mlp_probabilities
            ),
            "Training Time (seconds)": round(elapsed, 2),
            "Best Parameters": "hidden_layer_sizes=(32,16)"
        })

        if mlp_f2 > best_model_selection_score:
            best_model = mlp_pipeline
            best_model_name = "Neural Network (MLP)"
            best_model_selection_score = mlp_f2

        print("Validation F2 at 0.5:", round(mlp_f2, 4))
        print("Training time:", round(elapsed, 2), "seconds")

    print("\nSelected model:", best_model_name)


# =============================================================================
# 17. THRESHOLD OPTIMIZATION USING VALIDATION DATA ONLY
# =============================================================================

# A threshold below 0.5 can improve recall.
# F2 gives recall greater weight than precision.
# The test set is not used to select this threshold.

if not cached_model_loaded:

    val_probabilities = best_model.predict_proba(X_val)[:, 1]

    precision_values, recall_values, thresholds = (
        precision_recall_curve(y_val, val_probabilities)
    )

    # Precision/recall arrays have one more entry than thresholds.
    f2_values = (
        5 * precision_values[:-1] * recall_values[:-1]
        / (
            4 * precision_values[:-1]
            + recall_values[:-1]
            + 1e-12
        )
    )

    if len(thresholds) > 0:
        best_threshold_index = int(np.nanargmax(f2_values))
        best_threshold = float(thresholds[best_threshold_index])
    else:
        best_threshold = 0.5

    print("\nValidation-optimized threshold:", best_threshold)

else:
    print("\nUsing threshold loaded from Drive:", best_threshold)


# =============================================================================
# 18. FINAL TEST-SET EVALUATION
# =============================================================================

print("\n" + "=" * 75)
print("FINAL TEST-SET EVALUATION")
print("=" * 75)

test_probabilities = best_model.predict_proba(X_test)[:, 1]

test_predictions = (
    test_probabilities >= best_threshold
).astype(int)

test_predictions_05 = (
    test_probabilities >= 0.5
).astype(int)

def calculate_metrics(y_true, y_pred, y_prob):

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_true, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_true, y_pred, zero_division=0
        ),
        "F2 Score": fbeta_score(
            y_true, y_pred, beta=2, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_true, y_prob),
        "PR-AUC": average_precision_score(y_true, y_prob)
    }

final_metrics = calculate_metrics(
    y_test,
    test_predictions,
    test_probabilities
)

default_threshold_metrics = calculate_metrics(
    y_test,
    test_predictions_05,
    test_probabilities
)

print("\nSelected model:", best_model_name)
print("Decision threshold:", round(best_threshold, 5))

print("\nTest metrics at optimized threshold:")
display(
    pd.DataFrame(
        final_metrics,
        index=["Test Set"]
    ).T
)

print("\nTest metrics at default threshold 0.5:")
display(
    pd.DataFrame(
        default_threshold_metrics,
        index=["Test Set"]
    ).T
)

print("\nClassification report:")
print(
    classification_report(
        y_test,
        test_predictions,
        target_names=["No Diabetes", "Diabetes"],
        zero_division=0
    )
)

cm = confusion_matrix(y_test, test_predictions)

print("\nConfusion matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("\nConfusion matrix interpretation:")
print("True Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)

print("\nFalse negatives are particularly important in a screening context.")
print("However, reducing false negatives generally increases false positives.")


# =============================================================================
# 19. CONFUSION MATRIX VISUALIZATION
# =============================================================================

plt.figure(figsize=(7, 6))
plt.imshow(cm)
plt.colorbar(label="Number of cases")

plt.xticks([0, 1], ["Predicted No", "Predicted Yes"])
plt.yticks([0, 1], ["Actual No", "Actual Yes"])

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            str(cm[i, j]),
            ha="center",
            va="center",
            fontsize=13
        )

plt.title(f"Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "10_confusion_matrix.png"), dpi=150)
plt.show()


# =============================================================================
# 20. ROC CURVE AND PRECISION-RECALL CURVE
# =============================================================================

fpr, tpr, roc_thresholds = roc_curve(
    y_test,
    test_probabilities
)

plt.figure(figsize=(7, 6))
plt.plot(
    fpr,
    tpr,
    label=f"ROC-AUC = {final_metrics['ROC-AUC']:.4f}"
)
plt.plot([0, 1], [0, 1], linestyle="--")
plt.title("Receiver Operating Characteristic (ROC) Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "11_roc_curve.png"), dpi=150)
plt.show()

pr_precision, pr_recall, pr_thresholds = precision_recall_curve(
    y_test,
    test_probabilities
)

plt.figure(figsize=(7, 6))
plt.plot(
    pr_recall,
    pr_precision,
    label=f"PR-AUC = {final_metrics['PR-AUC']:.4f}"
)
plt.title("Precision-Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "12_precision_recall_curve.png"), dpi=150)
plt.show()


# =============================================================================
# 21. THRESHOLD ANALYSIS
# =============================================================================

threshold_grid = np.linspace(0.05, 0.95, 37)

threshold_results = []

for threshold in threshold_grid:

    pred = (test_probabilities >= threshold).astype(int)

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_test, pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, pred, zero_division=0
        ),
        "F1": f1_score(
            y_test, pred, zero_division=0
        ),
        "F2": fbeta_score(
            y_test, pred, beta=2, zero_division=0
        )
    })

threshold_df = pd.DataFrame(threshold_results)

plt.figure(figsize=(9, 6))
plt.plot(
    threshold_df["Threshold"],
    threshold_df["Precision"],
    label="Precision"
)
plt.plot(
    threshold_df["Threshold"],
    threshold_df["Recall"],
    label="Recall"
)
plt.plot(
    threshold_df["Threshold"],
    threshold_df["F1"],
    label="F1"
)
plt.plot(
    threshold_df["Threshold"],
    threshold_df["F2"],
    label="F2"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label="Validation-selected threshold"
)

plt.title("Metric Variation with Classification Threshold")
plt.xlabel("Decision Threshold")
plt.ylabel("Metric Value")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "13_threshold_analysis.png"), dpi=150)
plt.show()

# Note:
# This plot describes the test-set trade-off after the threshold was fixed
# using validation data. Do not use it to choose a new threshold.


# =============================================================================
# 22. MODEL COMPARISON
# =============================================================================

comparison_df = pd.DataFrame(model_comparison)

if not comparison_df.empty:
    comparison_df = comparison_df.sort_values(
        by="Validation F2 at 0.5",
        ascending=False
    )

print("\nModel comparison:")
display(comparison_df)

comparison_df.to_csv(COMPARISON_PATH, index=False)


# =============================================================================
# 23. FEATURE IMPORTANCE USING PERMUTATION IMPORTANCE
# =============================================================================

print("\nCalculating permutation feature importance...")

# Permutation importance evaluates the change in model performance when
# a feature is shuffled. It measures predictive contribution, not causation.

importance_sample_size = min(3000, len(X_val))

X_importance = X_val.sample(
    importance_sample_size,
    random_state=RANDOM_STATE
)

y_importance = y_val.loc[X_importance.index]

importance_result = permutation_importance(
    best_model,
    X_importance,
    y_importance,
    scoring="average_precision",
    n_repeats=5,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": FEATURE_COLUMNS,
    "Importance Mean": importance_result.importances_mean,
    "Importance Std": importance_result.importances_std
}).sort_values(
    "Importance Mean",
    ascending=False
)

print("\nFeature importance:")
display(importance_df)

importance_df.to_csv(FEATURE_IMPORTANCE_PATH, index=False)

plt.figure(figsize=(9, 6))
plt.barh(
    importance_df["Feature"],
    importance_df["Importance Mean"],
    xerr=importance_df["Importance Std"]
)
plt.gca().invert_yaxis()
plt.title("Permutation Feature Importance")
plt.xlabel("Decrease in Average Precision when Feature is Shuffled")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "14_feature_importance.png"), dpi=150)
plt.show()


# =============================================================================
# 24. FAIRNESS / SUBGROUP PERFORMANCE AUDIT
# =============================================================================

# These are descriptive subgroup comparisons, not proof of fairness.
# Differences can reflect population composition, sample sizes or dataset bias.

audit_df = X_test.copy()
audit_df["actual"] = y_test.values
audit_df["predicted"] = test_predictions
audit_df["probability"] = test_probabilities

audit_results = []

def add_audit_row(group_name, group_value, subset):

    if len(subset) == 0:
        return

    audit_results.append({
        "Group Feature": group_name,
        "Group": str(group_value),
        "Sample Size": len(subset),
        "Actual Positive Cases": int(subset["actual"].sum()),
        "Recall": recall_score(
            subset["actual"],
            subset["predicted"],
            zero_division=0
        ),
        "Precision": precision_score(
            subset["actual"],
            subset["predicted"],
            zero_division=0
        ),
        "False Negative Count": int(
            (
                (subset["actual"] == 1)
                & (subset["predicted"] == 0)
            ).sum()
        )
    })

for value, subset in audit_df.groupby("gender", dropna=False):
    add_audit_row("Gender", value, subset)

audit_df["age_group"] = pd.cut(
    audit_df["age"],
    bins=[-np.inf, 30, 45, 60, np.inf],
    labels=["Below 30", "30-45", "45-60", "Above 60"]
)

for value, subset in audit_df.groupby("age_group", observed=False):
    add_audit_row("Age Group", value, subset)

fairness_df = pd.DataFrame(audit_results)

print("\nSubgroup performance:")
display(fairness_df)


# =============================================================================
# 25. GENERALIZATION / OVERFITTING CHECK
# =============================================================================

train_probabilities = best_model.predict_proba(X_train)[:, 1]
train_predictions = (
    train_probabilities >= best_threshold
).astype(int)

train_metrics = calculate_metrics(
    y_train,
    train_predictions,
    train_probabilities
)

generalization_df = pd.DataFrame({
    "Training": train_metrics,
    "Test": final_metrics
})

generalization_df["Train-Test Difference"] = (
    generalization_df["Training"] - generalization_df["Test"]
)

print("\nGeneralization comparison:")
display(generalization_df)

print(
    "\nA large training-test performance gap may indicate overfitting. "
    "A small gap is not by itself proof of good generalization."
)


# =============================================================================
# 26. PCA: DIMENSIONALITY REDUCTION
# =============================================================================

print("\nPerforming PCA...")

# Fit the preprocessing transformation using training data only.
# Then fit PCA using the transformed training data only.

fitted_preprocessor = best_model.named_steps["prep"]

X_train_transformed = fitted_preprocessor.transform(X_train)

pca_full = PCA()
pca_full.fit(X_train_transformed)

cumulative_variance = np.cumsum(
    pca_full.explained_variance_ratio_
)

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA - Cumulative Explained Variance")
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "15_pca_explained_variance.png"), dpi=150)
plt.show()

pca_2d = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

X_train_pca = pca_2d.fit_transform(X_train_transformed)

print(
    "Variance explained by first two principal components:",
    round(pca_2d.explained_variance_ratio_.sum() * 100, 2),
    "%"
)

pca_sample_size = min(5000, len(X_train_pca))
pca_indices = np.random.choice(
    len(X_train_pca),
    size=pca_sample_size,
    replace=False
)

plt.figure(figsize=(8, 6))

pca_labels = y_train.iloc[pca_indices].values

for target_value, label in [(0, "No Diabetes"), (1, "Diabetes")]:
    mask = pca_labels == target_value

    plt.scatter(
        X_train_pca[pca_indices][mask, 0],
        X_train_pca[pca_indices][mask, 1],
        alpha=0.4,
        s=12,
        label=label
    )

plt.title("PCA Projection of Training Data")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "16_pca_projection.png"), dpi=150)
plt.show()


# =============================================================================
# 27. K-MEANS CLUSTERING AND WCSS
# =============================================================================

print("\nPerforming K-Means clustering...")

# K-Means is unsupervised: target labels are not used for fitting.
# We use a sample to keep execution practical.

cluster_sample_size = min(15000, len(X_train_transformed))

cluster_indices = np.random.choice(
    len(X_train_transformed),
    size=cluster_sample_size,
    replace=False
)

X_cluster = X_train_transformed[cluster_indices]

wcss_values = []

k_values = range(1, 9)

for k in k_values:

    km = MiniBatchKMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        batch_size=1024,
        n_init=5
    )

    km.fit(X_cluster)
    wcss_values.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(
    list(k_values),
    wcss_values,
    marker="o"
)
plt.title("K-Means Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS / Inertia")
plt.xticks(list(k_values))
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "17_kmeans_elbow.png"), dpi=150)
plt.show()

# A simple illustrative clustering with K=3.
kmeans_final = MiniBatchKMeans(
    n_clusters=3,
    random_state=RANDOM_STATE,
    batch_size=1024,
    n_init=5
)

cluster_labels = kmeans_final.fit_predict(X_cluster)

print("\nK-Means cluster sizes:")
display(
    pd.Series(cluster_labels)
    .value_counts()
    .sort_index()
    .to_frame("Number of Records")
)

# Visualize clusters in the first two PCA dimensions.
X_cluster_pca = PCA(
    n_components=2,
    random_state=RANDOM_STATE
).fit_transform(X_cluster)

plt.figure(figsize=(8, 6))
plt.scatter(
    X_cluster_pca[:, 0],
    X_cluster_pca[:, 1],
    c=cluster_labels,
    alpha=0.45,
    s=10
)
plt.title("K-Means Clusters Visualized in 2D PCA Space")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "18_kmeans_pca_clusters.png"), dpi=150)
plt.show()


# =============================================================================
# 28. SAVE THE TRAINED MODEL AND METADATA
# =============================================================================

# Save the fitted preprocessing + classifier pipeline.
joblib.dump(best_model, MODEL_PATH)

metadata = {
    "fingerprint": fingerprint,
    "data_hash": data_hash,
    "config_hash": config_hash,
    "best_model_name": best_model_name,
    "best_threshold": float(best_threshold),
    "random_state": RANDOM_STATE,
    "features": FEATURE_COLUMNS,
    "numeric_features": NUMERIC_FEATURES,
    "categorical_features": CATEGORICAL_FEATURES,
    "dataset_rows_after_cleaning": int(len(df)),
    "train_rows": int(len(X_train)),
    "validation_rows": int(len(X_val)),
    "test_rows": int(len(X_test)),
    "final_test_metrics": {
        key: float(value)
        for key, value in final_metrics.items()
    },
    "model_comparison": model_comparison
}

with open(METADATA_PATH, "w") as f:
    json.dump(metadata, f, indent=4)

pd.DataFrame(
    [final_metrics]
).to_csv(METRICS_PATH, index=False)

print("\nModel saved:", MODEL_PATH)
print("Metadata saved:", METADATA_PATH)
print("Metrics saved:", METRICS_PATH)


# =============================================================================
# 29. SAVE PROJECT SUMMARY
# =============================================================================

project_summary = {
    "project_title": "Diabetes Risk Classification",
    "institution": "NITK Surathkal",
    "course": "Data Science and Artificial Intelligence",
    "dataset_source": "Kaggle - iammustafatz/diabetes-prediction-dataset",
    "dataset_rows_after_cleaning": int(len(df)),
    "features": FEATURE_COLUMNS,
    "selected_model": best_model_name,
    "selected_threshold": float(best_threshold),
    "test_metrics": {
        key: float(value)
        for key, value in final_metrics.items()
    },
    "confusion_matrix": cm.tolist(),
    "true_negatives": int(tn),
    "false_positives": int(fp),
    "false_negatives": int(fn),
    "true_positives": int(tp),
    "important_features": importance_df.head(5)["Feature"].tolist(),
    "limitations": [
        "The model is trained on a Kaggle dataset and may not generalize "
        "to all real-world populations.",
        "The model is an educational screening demonstration, not a diagnosis.",
        "Threshold choice changes the balance between false negatives "
        "and false positives.",
        "Subgroup metrics can be unstable for small groups.",
        "Permutation importance indicates predictive contribution, "
        "not causation."
    ]
}

with open(SUMMARY_PATH, "w") as f:
    json.dump(project_summary, f, indent=4)

print("Project summary saved:", SUMMARY_PATH)


# =============================================================================
# 30. INTERACTIVE LIVE DEMONSTRATION WIDGET
# =============================================================================

import ipywidgets as widgets
from IPython.display import display, HTML

print("\n" + "=" * 75)
print("INTERACTIVE DIABETES RISK DEMONSTRATION")
print("=" * 75)

age_widget = widgets.FloatSlider(
    value=45,
    min=1,
    max=100,
    step=1,
    description="Age:",
    continuous_update=False
)

gender_widget = widgets.Dropdown(
    options=["Female", "Male"],
    value="Female",
    description="Gender:"
)

hypertension_widget = widgets.Dropdown(
    options=[("No", 0), ("Yes", 1)],
    value=0,
    description="Hypertension:"
)

heart_widget = widgets.Dropdown(
    options=[("No", 0), ("Yes", 1)],
    value=0,
    description="Heart disease:"
)

smoking_widget = widgets.Dropdown(
    options=["never", "former", "current"],
    value="never",
    description="Smoking:"
)

bmi_widget = widgets.FloatSlider(
    value=25.0,
    min=10,
    max=60,
    step=0.1,
    description="BMI:",
    continuous_update=False
)

hba1c_widget = widgets.FloatSlider(
    value=5.5,
    min=3,
    max=15,
    step=0.1,
    description="HbA1c:",
    continuous_update=False
)

glucose_widget = widgets.FloatSlider(
    value=110,
    min=50,
    max=400,
    step=1,
    description="Glucose:",
    continuous_update=False
)

threshold_widget = widgets.FloatSlider(
    value=float(best_threshold),
    min=0.01,
    max=0.99,
    step=0.01,
    description="Threshold:",
    readout_format=".2f",
    continuous_update=False
)

predict_button = widgets.Button(
    description="Predict Diabetes Risk",
    button_style="success",
    icon="check"
)

reset_button = widgets.Button(
    description="Reset Inputs",
    button_style="",
    icon="refresh"
)

output_widget = widgets.Output()

def make_input_row():

    return pd.DataFrame([{
        "gender": gender_widget.value,
        "age": float(age_widget.value),
        "hypertension": int(hypertension_widget.value),
        "heart_disease": int(heart_widget.value),
        "smoking_history": smoking_widget.value,
        "bmi": float(bmi_widget.value),
        "HbA1c_level": float(hba1c_widget.value),
        "blood_glucose_level": float(glucose_widget.value)
    }], columns=FEATURE_COLUMNS)


def predict_live(_):

    with output_widget:
        clear_output(wait=True)

        input_data = make_input_row()

        probability = float(
            best_model.predict_proba(input_data)[0, 1]
        )

        threshold = float(threshold_widget.value)
        prediction = int(probability >= threshold)

        print("MODEL:", best_model_name)
        print("PREDICTED PROBABILITY:", f"{probability * 100:.2f}%")
        print("DECISION THRESHOLD:", f"{threshold:.2f}")
        print()

        if prediction == 1:
            display(HTML(
                "<h3 style='color:#d62728'>"
                "Model classification: Higher predicted risk"
                "</h3>"
            ))
        else:
            display(HTML(
                "<h3 style='color:#2ca02c'>"
                "Model classification: Lower predicted risk"
                "</h3>"
            ))

        print("Input data:")
        display(input_data)

        print(
            "\nImportant: This is an educational machine-learning "
            "screening demonstration, not a medical diagnosis. "
            "A real health assessment requires a qualified healthcare "
            "professional and appropriate clinical testing."
        )


def reset_inputs(_):

    age_widget.value = 45
    gender_widget.value = "Female"
    hypertension_widget.value = 0
    heart_widget.value = 0
    smoking_widget.value = "never"
    bmi_widget.value = 25.0
    hba1c_widget.value = 5.5
    glucose_widget.value = 110
    threshold_widget.value = float(best_threshold)

    with output_widget:
        clear_output()


predict_button.on_click(predict_live)
reset_button.on_click(reset_inputs)

display(widgets.VBox([
    widgets.HTML("<h3>Enter Example Patient Information</h3>"),
    age_widget,
    gender_widget,
    hypertension_widget,
    heart_widget,
    smoking_widget,
    bmi_widget,
    hba1c_widget,
    glucose_widget,
    widgets.HTML("<h4>Decision Threshold</h4>"),
    threshold_widget,
    widgets.HBox([predict_button, reset_button]),
    output_widget
]))


# =============================================================================
# 31. FINAL PROJECT OUTPUTS AND COMPLETION SUMMARY
# =============================================================================

print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED")
print("=" * 75)

print("\nSELECTED MODEL:", best_model_name)
print("OPTIMIZED VALIDATION THRESHOLD:", round(best_threshold, 5))

print("\nFINAL TEST RESULTS:")
for metric, value in final_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nCONFUSION MATRIX:")
print(cm)

print("\nDRIVE OUTPUT LOCATION:")
print(PROJECT_DIR)

print("\nIMPORTANT SAVED FILES:")
for path in [
    CSV_PATH,
    CLEANED_DATA_PATH,
    MODEL_PATH,
    METADATA_PATH,
    METRICS_PATH,
    COMPARISON_PATH,
    FEATURE_IMPORTANCE_PATH,
    SUMMARY_PATH
]:
    print(" -", path)

print("\nVISUALIZATIONS FOLDER:")
print(PLOT_DIR)

print("\nTo avoid retraining next time:")
print("1. Keep FORCE_RETRAIN = False.")
print("2. Run this cell again.")
print("3. The saved model will load if the dataset and configuration match.")

print("\nTo train again from scratch:")
print("1. Change FORCE_RETRAIN = True.")
print("2. Run the cell.")
print("3. Change it back to False after training if desired.")
```
